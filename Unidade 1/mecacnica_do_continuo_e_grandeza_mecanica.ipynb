1. Representação simbólica

Utilizando SymPy, represente os pontos A=(2,1,3), B=(−1,2,4)

```python
x1, x2, x3 = sp.symbols("x_1 x_2 x_3")

x = sp.Matrix([x1, x2, x3])
x
```

```python
PontoA = x.subs({
    x1: 2,
    x2: 1,
    x3: 3
})

PontoB = x.subs({
    x1: -1,
    x2: 2,
    x3: 4
})

ponto = input("Qual ponto você quer ver? (A ou B): ")

if ponto == "a" or ponto == "A":
    display(PontoA)
elif ponto == "b" or ponto == "B":
    display(PontoB)
else:
    print("Ponto inválido. Por favor, escolha A ou B.")
```

3. Campo vetorial

Considere

v(x,y)=[x−y].

Represente graficamente esse campo utilizando quiver.

Compare o resultado com o campo apresentado no notebook.

```python
# Cria 15 pontos igualmente espaçados entre -2 e 2 para cada uma das coordenadas x e y
xv = np.linspace(-2, 2, 15)
yv = np.linspace(-2, 2, 15)

# Constrói uma malha bidimensional a partir dos valores de x e y
X, Y = np.meshgrid(xv, yv)

# Calcula as componentes do campo vetorial v(x,y) = (x, -y) em cada ponto da malha
U = X
V = -Y

fig, ax = plt.subplots(figsize=(6, 6))
ax.quiver(X, Y, U, V) #Desenha uma seta em cada ponto  (X,Y) (X,Y), apontando na direção  (U,V) (U,V)
ax.set_xlabel(r"$x$")
ax.set_ylabel(r"$y$")
ax.set_title(r"Campo vetorial $\mathbf{v}(x,y)=(x,-y)$")
ax.set_aspect("equal")
plt.show()
```
